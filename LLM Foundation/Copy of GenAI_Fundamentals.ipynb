## 🌱 What is Generative AI?

Generative AI refers to models that can **generate new content**:
- Text
- Images
- Code
- Audio

Unlike traditional ML (predict/classify), GenAI **creates**.

### Examples:
- ChatGPT → text generation
- DALL·E → image generation
- GitHub Copilot → code completion

## 🧠 Large Language Models (LLMs)

LLMs are deep learning models trained on massive text corpora.

Examples:
- BERT → Understanding-focused
- GPT → Generation-focused
- T5 → Text-to-text framework

Key properties:
- Billions of parameters
- Trained using self-supervised learning
- Use Transformer architecture

## 🔄 Transformer Architecture

Transformers replaced RNNs/LSTMs by enabling:
- Parallel processing
- Long-range dependency capture

### Core Components:
- Token Embeddings
- Positional Encoding
- Multi-Head Self-Attention
- Feed Forward Networks

## 👀 Attention Mechanism (Self-Attention)

Self-attention allows each word to **attend to all other words**.

It computes:
- Query (Q)
- Key (K)
- Value (V)

This helps the model understand **context** and **relationships**.

## 🔁 T5 (Text-to-Text Transfer Transformer)

T5 treats **every NLP task as text → text**:
- Translation
- Summarization
- QA

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("t5-small")

text = """
Artificial Intelligence is transforming industries by automating tasks,
improving decision-making, and enabling new applications.
"""

inputs = tokenizer(
    "summarize: " + text,
    return_tensors="pt",
    max_length=512,
    truncation=True
)

outputs = model.generate(
    **inputs,
    max_new_tokens=50
)

summary = tokenizer.decode(outputs[0], skip_special_tokens=True)

print(summary)
```

## ⚠️ Model Limitations – Hallucinations

LLMs may:
- Generate incorrect facts
- Sound confident but be wrong

Mitigation:
- Grounding with data
- Retrieval-Augmented Generation (RAG)
- Explicit constraints in prompts

```python
from transformers import pipeline

# 1. Sentiment analysis on 5 customer reviews using BERT
sentiment_analyzer = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

reviews = [
    "The battery life is incredible, lasts all day!",
    "Extremely disappointed with the build quality, it feels cheap.",
    "Decent performance for the price, but the camera is mediocre.",
    "Absolutely love this! Best purchase I've made this year.",
    "The customer service was rude and didn't help at all."
]

results = sentiment_analyzer(reviews)

for review, result in zip(reviews, results):
    print(f"Review: {review}\nSentiment: {result['label']} (Score: {result['score']:.4f})\n")
```

```python
import os
from groq import Groq
from dotenv import load_dotenv

load_dotenv()
client = Groq(api_key=os.getenv("GROQ_API_KEY"))

# 2. Generate a product description using Llama
def generate_description(product_name, key_features):
    prompt = f"Write a compelling product description for '{product_name}' with these features: {key_features}."
    
    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{"role": "user", "content": prompt}]
    )
    return response.choices[0].message.content

description = generate_description("EcoCharge Solar Power Bank", "20000mAh, waterproof, dual USB ports, fast charging")
print("Product Description:\n", description)
```

```python
# 3. Summarize a news article using T5 (GPT-OSS equivalent for summarization)
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("t5-small")
article = """
SpaceX successfully launched another batch of Starlink satellites into orbit on Wednesday, 
marking the company's 25th launch of the year. The Falcon 9 rocket lifted off from Cape Canaveral, 
carrying 23 satellites to join the growing constellation designed to provide global high-speed internet. 
The first-stage booster successfully landed on a drone ship in the Atlantic Ocean, 
demonstrating continued progress in rocket reusability.
"""

inputs = tokenizer(
    "summarize: " + article,
    return_tensors="pt",
    max_length=512,
    truncation=True
)

outputs = model.generate(
    **inputs,
    max_new_tokens=50
)

summary = tokenizer.decode(outputs[0], skip_special_tokens=True)

print(summary)
```

```python
# 4. Try zero-shot vs few-shot prompting
def compare_prompting(task_input):
    # Zero-shot
    zero_shot_prompt = f"Classify the sentiment of this text: '{task_input}'"
    
    # Few-shot
    few_shot_prompt = f"""
    Text: 'The movie was a masterpiece.' Sentiment: Positive
    Text: 'I hated every minute of it.' Sentiment: Negative
    Text: '{task_input}' Sentiment:
    """
    
    outputs = []
    for p in [zero_shot_prompt, few_shot_prompt]:
        res = client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            messages=[{"role": "user", "content": p}]
        )
        outputs.append(res.choices[0].message.content)
    
    print(f"Zero-shot Output:\n{outputs[0]}\n")
    print(f"Few-shot Output:\n{outputs[1]}")

compare_prompting("The experience was okay, nothing special but not bad either.")
```

```python
# 5. Identify hallucinations
prompt = "Who is the current world record holder for eating the most number of Mars bars in 2 minutes? Provide details."

response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=[{"role": "user", "content": prompt}]
)

print("Model Output:\n", response.choices[0].message.content)
print("\n--- Analysis ---")
print("To identify hallucinations, verify if the record exists in official sources like Guinness World Records.")
print("LLMs often hallucinate specific names or numbers for niche or non-existent records.")
```
